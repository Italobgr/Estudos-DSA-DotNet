# Desafios de Código Iniciais Mais Famosos

Os desafios de código iniciais são fundamentais para testar a lógica de programação, o conhecimento de estruturas de dados e a fluência na linguagem de programação de um candidato ou estudante. Eles frequentemente aparecem em entrevistas de emprego e plataformas de programação competitiva (como LeetCode, HackerRank, CodeSignal).

Abaixo estão detalhados alguns dos desafios mais famosos, o que eles avaliam e como abordá-los.

---

## 1. FizzBuzz
**O Problema:** 
Escreva um programa que imprima os números de 1 a 100. Mas, para múltiplos de 3, imprima "Fizz" em vez do número, e para múltiplos de 5, imprima "Buzz". Para números que são múltiplos de ambos (3 e 5), imprima "FizzBuzz".

**O que avalia:**
- Controle de fluxo (`if`, `else if`, `else`).
- Operador de módulo (`%`) para verificar a divisibilidade.
- Ordem de avaliação (verificar `n % 15 == 0` antes de `n % 3` ou `n % 5`).

**Como abordar:**
Sempre comece verificando a condição mais restritiva (múltiplo de 3 e 5, ou `n % 15 == 0`). Se essa condição for verdadeira, não precisamos verificar as outras individualmente. É um ótimo exercício inicial para garantir que o programador entende fluxo de decisão.

## 2. Two Sum (Soma de Dois Números)
**O Problema:**
Dado um array de números inteiros e um número alvo (`target`), retorne os **índices** dos dois números de forma que a soma deles seja igual ao alvo. Você pode assumir que cada entrada terá exatamente uma solução e não poderá usar o mesmo elemento duas vezes.

**O que avalia:**
- Uso de Estruturas de Dados eficientes.
- Troca de complexidade de tempo por espaço (Time-Space Tradeoff).

**Como abordar:**
- **Abordagem Ingênua:** Usar dois loops aninhados (força bruta). Isso dá uma complexidade de tempo de `O(n²)`.
- **Abordagem Otimizada:** Usar um Dicionário (`Dictionary` ou `Hash Map`). Enquanto iteramos pelo array, verificamos se o complemento (`target - elemento_atual`) já está no dicionário. Isso reduz o tempo para `O(n)`, mas usa espaço `O(n)`.

## 3. Verificação de Palíndromo
**O Problema:**
Dada uma string, verifique se ela é um palíndromo (se lê da mesma forma de trás para frente, ignorando espaços e diferenciação de maiúsculas e minúsculas).

**O que avalia:**
- Manipulação de strings.
- Uso da técnica de dois ponteiros (Two Pointers).

**Como abordar:**
Você pode colocar dois ponteiros: um no início da string e outro no final. Compare os caracteres. Se eles forem iguais, mova os ponteiros para o centro. Ignore caracteres não alfanuméricos durante o processo. A complexidade de tempo dessa abordagem é `O(n)` e a complexidade de espaço é `O(1)`.

## 4. Fatorial e Sequência de Fibonacci
**O Problema:**
- **Fatorial:** Calcular o produto de todos os números inteiros positivos menores ou iguais a `n`.
- **Fibonacci:** Retornar o `n-ésimo` número na sequência de Fibonacci, onde cada número é a soma dos dois anteriores (0, 1, 1, 2, 3, 5...).

**O que avalia:**
- Compreensão de **Recursão** vs **Iteração**.
- Problemas de performance com recursões simples e a necessidade de **Memoization** ou Programação Dinâmica.

**Como abordar:**
A solução recursiva pura de Fibonacci tem complexidade `O(2^n)` (exponencial e ineficiente). A melhor forma de resolver no início é usar um loop simples guardando os dois últimos valores ou usar um array (complexidade de tempo `O(n)`).

## 5. Anagrama Válido
**O Problema:**
Dadas duas strings, verifique se a segunda é um anagrama da primeira (se usa exatamente as mesmas letras, na mesma quantidade, mas em ordem diferente).

**O que avalia:**
- Contagem de frequências de caracteres.
- Ordenação de strings ou uso de `Hash Maps` / arrays de contagem.

**Como abordar:**
Uma forma simples é ordenar as duas strings e comparar se são iguais (`O(n log n)`). Uma solução melhor e mais rápida (`O(n)`) é criar um array de 26 posições (para o alfabeto) e incrementar a contagem para as letras da primeira string, e decrementar para as letras da segunda string. No final, todos os valores do array devem ser 0.

## Por que esses desafios importam?
Eles parecem simples, mas oferecem muito espaço para discussão em uma entrevista. Um bom engenheiro não apenas resolve o problema, mas discute:
1. **Casos Extremos (Edge cases):** Arrays vazios, números negativos, strings com 1 caractere, etc.
2. **Complexidade (Notação Big-O):** Quanto tempo de processamento e memória a solução consome.
3. **Escalabilidade:** O que acontece se o array não couber na memória?
