## 1. Agregações (Sum, Average, Min, Max)
Usados para realizar cálculos rápidos em coleções.

```polyglot-notebook
// Qual o valor total em estoque (soma de todos os preços)?
double valorTotal = produtos.Sum(p => p.Preco);
Console.WriteLine($"Valor total de todos os produtos: {valorTotal}");

// Qual o preço do produto mais caro e do mais barato?
double maisCaro = produtos.Max(p => p.Preco);
double maisBarato = produtos.Min(p => p.Preco);
Console.WriteLine($"Mais caro: {maisCaro} - Mais barato: {maisBarato}");

// Qual a média de preço dos eletrônicos?
double mediaEletronicos = produtos.Where(p => p.Categoria == "Eletrônicos").Average(p => p.Preco);
Console.WriteLine($"Média dos eletrônicos: {mediaEletronicos}");
```

## 2. Buscando Elementos Específicos (First vs FirstOrDefault)
- `First()`: Retorna o primeiro elemento que atende à condição. **Dá erro** se não achar nada.
- `FirstOrDefault()`: Retorna o primeiro elemento. **Retorna nulo** se não achar nada. (Recomendado)

```polyglot-notebook
// Pegar o primeiro produto que custa menos de 100
var produtoBaratinho = produtos.FirstOrDefault(p => p.Preco < 100);
Console.WriteLine($"Achei um baratinho: {produtoBaratinho?.Nome}");

// Tentar achar um Carro (não existe)
var carro = produtos.FirstOrDefault(p => p.Categoria == "Veículos");
if (carro == null)
{
    Console.WriteLine("Nenhum veículo encontrado na lista.");
}
```

## 3. Paginação (Skip e Take)
Muito usado para limitar resultados.
- `Skip(N)`: Pula os primeiros N registros.
- `Take(N)`: Pega apenas os próximos N registros.

```polyglot-notebook
// Imagine que estamos ordenando por preço e queremos os produtos da 2ª página (de 2 em 2)
// Pula os 2 primeiros mais baratos e pega os próximos 2
var segundaPagina = produtos.OrderBy(p => p.Preco).Skip(2).Take(2);

Console.WriteLine("Produtos da página 2:");
foreach(var p in segundaPagina)
{
    Console.WriteLine($"- {p.Nome} ({p.Preco})");
}
```
